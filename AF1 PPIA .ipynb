## Ejercicio 1: Evaluador de Rendimiento de Modelos



**Descripción de la lógica:**

Creamos la función `evaluar_modelo()`. Primero le pide al usuario la exactitud del modelo con `input()` y la convierte a número decimal con `float()`.

Después usamos `if`, `elif` y `else` para revisar en qué rango está el número y elegir la categoría:

- Mayor o igual a 0.90 → **Excelente**
- Entre 0.75 y 0.90 → **Aceptable**
- Entre 0.50 y 0.75 → **Requiere Ajuste**
- Menor a 0.50 → **Crítico / No apto**

Usamos el operador lógico `and` para revisar que el número esté entre dos valores. Al final la función imprime la categoría y la devuelve con `return`.


```python
def evaluar_modelo():
    exactitud = float(input("Ingresa la exactitud del modelo (entre 0.0 y 1.0): "))

    if exactitud >= 0.90:
        categoria = "Excelente"
    elif exactitud >= 0.75 and exactitud < 0.90:
        categoria = "Aceptable"
    elif exactitud >= 0.50 and exactitud < 0.75:
        categoria = "Requiere Ajuste"
    else:
        categoria = "Crítico / No apto"

    print("La categoría del modelo es:", categoria)
    return categoria

evaluar_modelo()
```

## Ejercicio 2: Procesamiento de Hiperparámetros y Lotes



**Descripción de la lógica:**

Primero creamos la tupla `CONFIGURACION_BASE` con los valores `(512, 10)`, que son las dimensiones de entrada y salida de una capa. Usamos una tupla porque sus valores no se pueden cambiar.

Después creamos la función `procesar_tasas_aprendizaje()`, que recibe una lista de tasas de aprendizaje. Dentro de la función:

1. Creamos una lista vacía llamada `tasas_escaladas`.
2. Con un bucle `for` recorremos cada tasa de la lista.
3. Multiplicamos cada tasa por 2.0 y la agregamos a la nueva lista con `append()`.
4. Imprimimos la nueva lista y la tupla, y las devolvemos con `return`.


```python
CONFIGURACION_BASE = (512, 10)

def procesar_tasas_aprendizaje(tasas):
    tasas_escaladas = []

    for tasa in tasas:
        nueva_tasa = tasa * 2.0
        tasas_escaladas.append(nueva_tasa)

    print("Tasas originales:", tasas)
    print("Tasas escaladas:", tasas_escaladas)
    print("Configuración base:", CONFIGURACION_BASE)

    return tasas_escaladas, CONFIGURACION_BASE

lista_tasas = [0.1, 0.01, 0.001, 0.0001]
procesar_tasas_aprendizaje(lista_tasas)
```

## Ejercicio 3: Simulador de Convergencia por Épocas



**Descripción de la lógica:**

Creamos la función `simular_entrenamiento()`. Primero le pide al usuario el número máximo de épocas con `input()` y lo convierte a número entero con `int()`.

Después usamos un bucle `for` con `range()` para recorrer las épocas desde la 1 hasta el número que dio el usuario.

En cada época calculamos la pérdida (*loss*) con la fórmula `perdida = 1 / epoca`. Así, mientras más avanza el entrenamiento, más pequeña es la pérdida (1.0, 0.5, 0.33, 0.25...).

Dentro del bucle usamos un `if` para revisar si la pérdida es menor a 0.05. Si eso pasa, imprimimos que el modelo convergió antes de tiempo y detenemos el bucle con `break`. El modelo converge en la época 21, así que si el usuario pide menos épocas el bucle termina sin llegar a converger.


```python
def simular_entrenamiento():
    max_epocas = int(input("Ingresa el número máximo de épocas: "))
    umbral = 0.05

    for epoca in range(1, max_epocas + 1):
        perdida = 1 / epoca
        print("Época", epoca, "- Pérdida:", round(perdida, 4))

        if perdida < umbral:
            print("El modelo ha convergido anticipadamente en la época", epoca)
            break

    print("Entrenamiento terminado.")
simular_entrenamiento()
```
